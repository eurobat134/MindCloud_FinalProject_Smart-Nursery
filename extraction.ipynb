```python
import os
import numpy as np
import pandas as pd
import librosa


# ============================================================
# SETTINGS
# ============================================================

dataset_folder = "csv audio"

output_csv = "cry_features.csv"

sample_rate = 16000


# ============================================================
# CRY REASON LABELS
# ============================================================

cry_reasons = {
    "discomfort": 0,
    "hungry": 1,
    "tired": 2
}


# ============================================================
# COLUMNS FOR THE CSV
# ============================================================

columns = [
    "Cry_Audio_File",
    "Cry_Reason",
    "Amplitude_Envelope_Mean",
    "RMS_Mean",
    "ZCR_Mean",
    "STFT_Mean",
    "SC_Mean",
    "SBAN_Mean",
    "SCON_Mean",
    "MFCCs13Mean",
    "delMFCCs13",
    "del2MFCCs13",
    "MelSpec",
    "MFCCs20",
    "MFCCs1",
    "MFCCs2",
    "MFCCs3",
    "MFCCs4",
    "MFCCs5",
    "MFCCs6",
    "MFCCs7",
    "MFCCs8",
    "MFCCs9",
    "MFCCs10",
    "MFCCs11",
    "MFCCs12",
    "MFCCs13"
]


# ============================================================
# STORE ALL FEATURES HERE
# ============================================================

data = []


# ============================================================
# GO THROUGH EACH FOLDER
# ============================================================

for cry_reason, label in cry_reasons.items():

    folder_path = os.path.join(dataset_folder, cry_reason)

    for filename in os.listdir(folder_path):

        if not filename.lower().endswith(".wav"):
            continue

        file_path = os.path.join(folder_path, filename)

        print("Processing:", file_path)


        # ====================================================
        # LOAD AUDIO
        # ====================================================

        audio, sr = librosa.load(
            file_path,
            sr=sample_rate,
            mono=True
        )


        # ====================================================
        # AMPLITUDE ENVELOPE MEAN
        # ====================================================

        amplitude_envelope = np.abs(audio)

        amplitude_envelope_mean = np.mean(
            amplitude_envelope
        )


        # ====================================================
        # RMS MEAN
        # ====================================================

        rms = librosa.feature.rms(
            y=audio
        )

        RMS_Mean = np.mean(rms)


        # ====================================================
        # ZCR MEAN
        # ====================================================

        zcr = librosa.feature.zero_crossing_rate(
            audio
        )

        ZCR_Mean = np.mean(zcr)


        # ====================================================
        # STFT MEAN
        # ====================================================

        stft = librosa.stft(audio)

        stft_magnitude = np.abs(stft)

        STFT_Mean = np.mean(
            stft_magnitude
        )


        # ====================================================
        # SC MEAN
        # SPECTRAL CENTROID
        # ====================================================

        spectral_centroid = librosa.feature.spectral_centroid(
            y=audio,
            sr=sr
        )

        SC_Mean = np.mean(
            spectral_centroid
        )


        # ====================================================
        # SBAN MEAN
        # SPECTRAL BANDWIDTH
        # ====================================================

        spectral_bandwidth = librosa.feature.spectral_bandwidth(
            y=audio,
            sr=sr
        )

        SBAN_Mean = np.mean(
            spectral_bandwidth
        )


        # ====================================================
        # SCON MEAN
        # SPECTRAL CONTRAST
        # ====================================================

        spectral_contrast = librosa.feature.spectral_contrast(
            y=audio,
            sr=sr
        )

        SCON_Mean = np.mean(
            spectral_contrast
        )


        # ====================================================
        # 13 MFCCs
        # ====================================================

        mfcc13 = librosa.feature.mfcc(
            y=audio,
            sr=sr,
            n_mfcc=13
        )


        # ====================================================
        # MFCCs13Mean
        # ====================================================

        MFCCs13Mean = np.mean(
            mfcc13
        )


        # ====================================================
        # DELTA MFCCs13
        # ====================================================

        delta_mfcc13 = librosa.feature.delta(
            mfcc13
        )

        delMFCCs13 = np.mean(
            delta_mfcc13[12]
        )


        # ====================================================
        # DELTA-DELTA MFCCs13
        # ====================================================

        delta2_mfcc13 = librosa.feature.delta(
            mfcc13,
            order=2
        )

        del2MFCCs13 = np.mean(
            delta2_mfcc13[12]
        )


        # ====================================================
        # MEL SPECTROGRAM
        # ====================================================

        mel_spec = librosa.feature.melspectrogram(
            y=audio,
            sr=sr
        )

        # Convert matrix into a string so it can be stored
        # inside a CSV cell.
        MelSpec = str(
            mel_spec.tolist()
        )


        # ====================================================
        # 20 MFCCs
        # ====================================================

        mfcc20 = librosa.feature.mfcc(
            y=audio,
            sr=sr,
            n_mfcc=20
        )

        # Convert matrix into a string for the CSV.
        MFCCs20 = str(
            mfcc20.tolist()
        )


        # ====================================================
        # INDIVIDUAL MFCCs 1-13
        # ====================================================

        MFCCs1 = np.mean(mfcc13[0])

        MFCCs2 = np.mean(mfcc13[1])

        MFCCs3 = np.mean(mfcc13[2])

        MFCCs4 = np.mean(mfcc13[3])

        MFCCs5 = np.mean(mfcc13[4])

        MFCCs6 = np.mean(mfcc13[5])

        MFCCs7 = np.mean(mfcc13[6])

        MFCCs8 = np.mean(mfcc13[7])

        MFCCs9 = np.mean(mfcc13[8])

        MFCCs10 = np.mean(mfcc13[9])

        MFCCs11 = np.mean(mfcc13[10])

        MFCCs12 = np.mean(mfcc13[11])

        MFCCs13 = np.mean(mfcc13[12])


        # ====================================================
        # PUT EVERYTHING INTO ONE ROW
        # ====================================================

        row = [
            file_path,
            label,

            amplitude_envelope_mean,
            RMS_Mean,
            ZCR_Mean,
            STFT_Mean,
            SC_Mean,
            SBAN_Mean,
            SCON_Mean,

            MFCCs13Mean,
            delMFCCs13,
            del2MFCCs13,

            MelSpec,
            MFCCs20,

            MFCCs1,
            MFCCs2,
            MFCCs3,
            MFCCs4,
            MFCCs5,
            MFCCs6,
            MFCCs7,
            MFCCs8,
            MFCCs9,
            MFCCs10,
            MFCCs11,
            MFCCs12,
            MFCCs13
        ]


        # Add row to dataset
        data.append(row)


# ============================================================
# CREATE DATAFRAME
# ============================================================

df = pd.DataFrame(
    data,
    columns=columns
)


# ============================================================
# SAVE CSV
# ============================================================

df.to_csv(
    output_csv,
    index=False
)


# ============================================================
# DONE
# ============================================================

print("\nFinished!")

print("Number of audio files:", len(df))

print("CSV saved as:", output_csv)
```

```python
import os
import numpy as np
import pandas as pd
import librosa

dataset_folder = "cry-not_cry audio"
output_csv = "cry_not_cry_features.csv"
sample_rate = 16000

cry_folder = "cry"
not_cry_folder = "not_cry"


def get_wav_files(folder_path):
    files = []

    for filename in os.listdir(folder_path):
        file_path = os.path.join(folder_path, filename)

        if os.path.isfile(file_path) and filename.lower().endswith((".wav", ".ogg")):
            files.append(file_path)

    return files


def extract_features(file_path, label):

    print("Processing:", file_path)

    audio, sr = librosa.load(
        file_path,
        sr=sample_rate,
        mono=True
    )

    amplitude_envelope = np.abs(audio)
    amplitude_envelope_mean = np.mean(amplitude_envelope)

    rms = librosa.feature.rms(y=audio)
    RMS_Mean = np.mean(rms)

    zcr = librosa.feature.zero_crossing_rate(audio)
    ZCR_Mean = np.mean(zcr)

    stft = librosa.stft(audio)
    stft_magnitude = np.abs(stft)
    STFT_Mean = np.mean(stft_magnitude)

    spectral_centroid = librosa.feature.spectral_centroid(
        y=audio,
        sr=sr
    )
    SC_Mean = np.mean(spectral_centroid)

    spectral_bandwidth = librosa.feature.spectral_bandwidth(
        y=audio,
        sr=sr
    )
    SBAN_Mean = np.mean(spectral_bandwidth)

    spectral_contrast = librosa.feature.spectral_contrast(
        y=audio,
        sr=sr
    )
    SCON_Mean = np.mean(spectral_contrast)

    mfcc13 = librosa.feature.mfcc(
        y=audio,
        sr=sr,
        n_mfcc=13
    )

    MFCCs13Mean = np.mean(mfcc13)

    delta_mfcc13 = librosa.feature.delta(mfcc13)
    delMFCCs13 = np.mean(delta_mfcc13[12])

    delta2_mfcc13 = librosa.feature.delta(
        mfcc13,
        order=2
    )
    del2MFCCs13 = np.mean(delta2_mfcc13[12])

    mel_spec = librosa.feature.melspectrogram(
        y=audio,
        sr=sr
    )
    MelSpec = str(mel_spec.tolist())

    mfcc20 = librosa.feature.mfcc(
        y=audio,
        sr=sr,
        n_mfcc=20
    )
    MFCCs20 = str(mfcc20.tolist())

    MFCCs1 = np.mean(mfcc13[0])
    MFCCs2 = np.mean(mfcc13[1])
    MFCCs3 = np.mean(mfcc13[2])
    MFCCs4 = np.mean(mfcc13[3])
    MFCCs5 = np.mean(mfcc13[4])
    MFCCs6 = np.mean(mfcc13[5])
    MFCCs7 = np.mean(mfcc13[6])
    MFCCs8 = np.mean(mfcc13[7])
    MFCCs9 = np.mean(mfcc13[8])
    MFCCs10 = np.mean(mfcc13[9])
    MFCCs11 = np.mean(mfcc13[10])
    MFCCs12 = np.mean(mfcc13[11])
    MFCCs13 = np.mean(mfcc13[12])

    return [
        file_path,
        label,
        amplitude_envelope_mean,
        RMS_Mean,
        ZCR_Mean,
        STFT_Mean,
        SC_Mean,
        SBAN_Mean,
        SCON_Mean,
        MFCCs13Mean,
        delMFCCs13,
        del2MFCCs13,
        MelSpec,
        MFCCs20,
        MFCCs1,
        MFCCs2,
        MFCCs3,
        MFCCs4,
        MFCCs5,
        MFCCs6,
        MFCCs7,
        MFCCs8,
        MFCCs9,
        MFCCs10,
        MFCCs11,
        MFCCs12,
        MFCCs13
    ]


columns = [
    "Audio_File",
    "Is_Cry",
    "Amplitude_Envelope_Mean",
    "RMS_Mean",
    "ZCR_Mean",
    "STFT_Mean",
    "SC_Mean",
    "SBAN_Mean",
    "SCON_Mean",
    "MFCCs13Mean",
    "delMFCCs13",
    "del2MFCCs13",
    "MelSpec",
    "MFCCs20",
    "MFCCs1",
    "MFCCs2",
    "MFCCs3",
    "MFCCs4",
    "MFCCs5",
    "MFCCs6",
    "MFCCs7",
    "MFCCs8",
    "MFCCs9",
    "MFCCs10",
    "MFCCs11",
    "MFCCs12",
    "MFCCs13"
]


cry_path = os.path.join(
    dataset_folder,
    cry_folder
)

not_cry_path = os.path.join(
    dataset_folder,
    not_cry_folder
)


cry_files = get_wav_files(cry_path)
not_cry_files = get_wav_files(not_cry_path)

print("\n=========================")
print("FILE COUNTS")
print("=========================")

print("Cry WAV files:", len(cry_files))
print("Not-Cry WAV files:", len(not_cry_files))
print("Total WAV files:", len(cry_files) + len(not_cry_files))


data = []
failed_files = []


# Cry = 1
for file_path in cry_files:

    try:
        row = extract_features(file_path, 1)
        data.append(row)

    except Exception as e:
        print("FAILED:", file_path)
        print("ERROR:", e)
        failed_files.append(file_path)


# Not-Cry = 0
for file_path in not_cry_files:

    try:
        row = extract_features(file_path, 0)
        data.append(row)

    except Exception as e:
        print("FAILED:", file_path)
        print("ERROR:", e)
        failed_files.append(file_path)


df = pd.DataFrame(
    data,
    columns=columns
)

df.to_csv(
    output_csv,
    index=False
)


print("\n=========================")
print("FINISHED")
print("=========================")

print("Cry files found:", len(cry_files))
print("Not-Cry files found:", len(not_cry_files))
print("Successfully processed:", len(df))
print("Failed files:", len(failed_files))

print("\nCSV distribution:")
print(df["Is_Cry"].value_counts().sort_index())

print("\nCSV saved as:")
print(output_csv)
```
