```python
import os
import librosa
import soundfile as sf
import numpy as np


input_folder = "donateacry_corpus"
output_folder = "csv audio"

speed_rates = np.arange(0.90, 1.101, 0.02)
pitch_steps = [-2, -1, 1, 2]

cry_types = ["discomfort", "tired"]


for cry_type in cry_types:

    input_path = os.path.join(input_folder, cry_type)
    output_path = os.path.join(output_folder, cry_type)

    os.makedirs(output_path, exist_ok=True)

    files = [
        f for f in os.listdir(input_path)
        if f.lower().endswith(".wav")
        and "_speed_" not in f
        and "_pitch_" not in f
    ]

    print(f"\n{cry_type}: {len(files)} original files")

    for file_number, filename in enumerate(files, 1):

        file_path = os.path.join(input_path, filename)

        audio, sr = librosa.load(
            file_path,
            sr=None
        )

        name = os.path.splitext(filename)[0]


        # SPEED

        for rate in speed_rates:

            if np.isclose(rate, 1.0):
                continue

            output_name = f"{name}_speed_{rate:.2f}.wav"
            output_file = os.path.join(output_path, output_name)

            if os.path.exists(output_file):
                continue

            augmented = librosa.effects.time_stretch(
                audio,
                rate=rate
            )

            sf.write(
                output_file,
                augmented,
                sr
            )


        # PITCH

        for steps in pitch_steps:

            output_name = f"{name}_pitch_{steps:+d}.wav"
            output_file = os.path.join(output_path, output_name)

            if os.path.exists(output_file):
                continue

            augmented = librosa.effects.pitch_shift(
                audio,
                sr=sr,
                n_steps=steps
            )

            sf.write(
                output_file,
                augmented,
                sr
            )


        print(f"{file_number}/{len(files)}: {filename}")


print("\nDone!")
```
